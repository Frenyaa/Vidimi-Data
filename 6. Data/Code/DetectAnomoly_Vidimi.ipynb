```python
fraud_cases = []

def jaccard(a, b):
    return len(a & b) / len(a | b) if (a | b) else 0

threshold = 0.9  # ngưỡng giống nhau

# Lặp từng user
for (user, day, place), subdf in grouped.groupby(["user_id", "Ngày_only", "Place ID"]):
    bills = subdf.to_dict("records")
    for i in range(len(bills)):
        for j in range(i+1, len(bills)):
            sim = jaccard(bills[i]["product_name"], bills[j]["product_name"])
            if sim > threshold:
                fraud_cases.append({
                    "user_id": user,
                    "Ngày": day,
                    "Place ID": place,
                    "bill_id1": bills[i]["bill_id"],
                    "bill_id2": bills[j]["bill_id"],
                    "similarity": sim
                })
```

```python
def detect_fraud(df, threshold=0.9, export_csv=False, csv_path="fraud_detected.csv"):
    """
    Detect gian lận hóa đơn dựa trên user_id, sản phẩm, ngày, place_id.
    Điều kiện:
        - Cùng user_id
        - Cùng ngày (YYYY-MM-DD)
        - Cùng Place ID
        - Tập sản phẩm trong bill giống nhau (theo Jaccard similarity > threshold)

    Parameters
    ----------
    df : pandas.DataFrame
        Yêu cầu có các cột:
            - "user_id"
            - "bill_id"
            - "product_name"
            - "Ngày" (datetime hoặc string)
            - "Place ID"
    threshold : float, optional
        Ngưỡng Jaccard similarity (0-1), mặc định = 0.9
    export_csv : bool, optional
        Nếu True thì export kết quả ra file CSV
    csv_path : str, optional
        Đường dẫn file csv xuất ra

    Returns
    -------
    fraud_df : pandas.DataFrame
        DataFrame chứa các cặp bill nghi ngờ gian lận
    """

    # --- 1. Chuẩn hóa cột Ngày ---
    df = df.copy()
    df["Ngày"] = pd.to_datetime(df["Ngày"])
    df["Ngày_only"] = df["Ngày"].dt.date

    # --- 2. Gom sản phẩm theo bill ---
    grouped = (
        df.groupby(["User ID", "Bill ID", "Ngày_only", "Place ID"])["Item Name"]
          .apply(set)  # tập sản phẩm để loại trùng trong 1 bill
          .reset_index()
    )

    # --- 3. Hàm tính Jaccard ---
    def jaccard(a, b):
        return len(a & b) / len(a | b) if (a | b) else 0

    # --- 4. So sánh trong từng (user, ngày, place) ---
    fraud_cases = []
    for (user, day, place), subdf in grouped.groupby(["User ID", "Ngày_only", "Place ID"]):
        bills = subdf.to_dict("records")
        for i in range(len(bills)):
            for j in range(i+1, len(bills)):
                sim = jaccard(bills[i]["Item Name"], bills[j]["Item Name"])
                if sim > threshold:
                    fraud_cases.append({
                        "User ID": user,
                        "Ngày_only": day,
                        "Place ID": place,
                        "bill_id1": bills[i]["Bill ID"],
                        "bill_id2": bills[j]["Bill ID"],
                        "similarity": sim
                    })

    fraud_df = pd.DataFrame(fraud_cases)

    # --- 5. Export nếu cần ---
    if export_csv and not fraud_df.empty:
      fraud_df.to_excel("fraud_cases.xlsx", index=False)
      print("Đã xuất file fraud_cases.xlsx")

    return fraud_df

```

```python
import pandas as pd

def detect_fraud(df, threshold=0.9, export_excel=False, excel_path="fraud_cases.xlsx"):

    # --- 1. Chuẩn hóa cột Ngày ---
    df = df.copy()
    df["Ngày"] = pd.to_datetime(df["Ngày"])
    df["Ngày_only"] = df["Ngày"].dt.date

    # --- 2. Gom sản phẩm theo bill ---
    grouped = (
        df.groupby(["User ID", "Bill ID", "Ngày_only", "Place ID"])["Item Name"]
          .apply(set)  # tập sản phẩm để loại trùng trong 1 bill
          .reset_index()
    )

    # --- 3. Hàm tính Jaccard ---
    def jaccard(a, b):
        return len(a & b) / len(a | b) if (a | b) else 0

    # --- 4. So sánh trong từng (user, ngày, place) ---
    fraud_groups = []
    for (user, day, place), subdf in grouped.groupby(["User ID", "Ngày_only", "Place ID"]):
        bills = subdf.to_dict("records")
        suspicious_bills = set()

        for i in range(len(bills)):
            for j in range(i+1, len(bills)):
                sim = jaccard(bills[i]["Item Name"], bills[j]["Item Name"])
                if sim > threshold:
                    suspicious_bills.add(bills[i]["Bill ID"])
                    suspicious_bills.add(bills[j]["Bill ID"])

        if suspicious_bills:  # chỉ lưu nếu có gian lận
            fraud_groups.append({
                "User ID": user,
                "Ngày_only": day,
                "Place ID": place,
                "bill_problem": list(suspicious_bills)  # danh sách bill
            })

    fraud_df = pd.DataFrame(fraud_groups)

    # --- 5. Export nếu cần ---
    if export_excel and not fraud_df.empty:
        fraud_df.to_excel(excel_path, index=False)
        print(f"Đã xuất file {excel_path}")

    return fraud_df
```

```python
result = detect_fraud(df, threshold=0.9, export_excel=True)
print(result)
```
