````python
def chunk_list(data, chunk_size):
    """Chia list thành các chunk có kích thước chunk_size"""
    for i in range(0, len(data), chunk_size):
        yield data[i:i + chunk_size]
````

````python
# List to store the model's responses
gemini_responses = []
````

gemini_responses

````python
import datetime
import time

# Nhập phạm vi chunk muốn chạy (tính từ 1 cho dễ nhìn)
start_chunk = int(input("Nhập số chunk bắt đầu: "))
end_chunk = int(input("Nhập số chunk kết thúc: "))

# Đảm bảo không vượt ngoài phạm vi
start_chunk = max(1, start_chunk)
end_chunk = min(len(formatted_data_chunks), end_chunk)

# Tính số lượng dữ liệu trong 1 chunk
chunk_size = len(formatted_data_chunks[0]) if formatted_data_chunks else 0

# Tạo tên file với timestamp và chunk size
timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
log_filename = f"gemini_log_chunk{chunk_size}_range{start_chunk}-{end_chunk}_{timestamp}.txt"

# Mở file để ghi log
with open(log_filename, 'w', encoding='utf-8') as log_file:
    # Ghi header
    log_file.write(f"=== Gemini API Processing Log ===\n")
    log_file.write(f"Start time: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
    log_file.write(f"Chunk size: {chunk_size} items\n")
    log_file.write(f"Processing range: Chunk {start_chunk} to {end_chunk}\n")
    log_file.write(f"Total chunks: {len(formatted_data_chunks)}\n")
    log_file.write("=" * 50 + "\n\n")

    # Duyệt qua range cần chạy (chú ý: index trong Python bắt đầu từ 0)
    for i in range(start_chunk-1, end_chunk):
        try:
            status_msg = f"Processing chunk {i+1}/{len(formatted_data_chunks)}"
            print(status_msg)
            log_file.write(status_msg + "\n")

            # Combine the prompt and the data chunk
            full_prompt = prompt + "\n".join(formatted_data_chunks[i])

            # Gọi API Gemini mô hình 2.0
            response = client.models.generate_content(
                model="gemini-2.0-flash-lite",
                contents=full_prompt,
                config=types.GenerateContentConfig(
                    thinking_config=types.ThinkingConfig(thinking_budget=0)
                ),
            )

            # Append the model's response to the list
            gemini_responses.append(response.text)

            # Chỉ in ra response của chunk hiện tại
            response_header = f"\n--- Response for chunk {i+1} ---"
            print(response_header)
            print(response.text)
            print("-" * 20)

            # Ghi vào log file
            log_file.write(response_header + "\n")
            log_file.write(response.text + "\n")
            log_file.write("-" * 20 + "\n")

            # Pause 60s before next request
            if i < end_chunk - 1:
                wait_msg = "⏳ Waiting 60s to avoid rate limit..."
                print(wait_msg)
                log_file.write(wait_msg + "\n\n")
                log_file.flush()  # Đảm bảo ghi ngay vào file
                time.sleep(60)

        except Exception as e:
            error_msg = f"Error processing chunk {i+1}: {e}"
            print(error_msg)
            log_file.write(error_msg + "\n\n")
            gemini_responses.append(error_msg)

    # Ghi footer
    log_file.write("\n" + "=" * 50 + "\n")
    log_file.write(f"End time: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
    log_file.write(f"Total chunks processed: {end_chunk - start_chunk + 1}\n")
    log_file.write(f"Log saved to: {log_filename}\n")

print(f"\n✅ Log file saved: {log_filename}")
````

````python
def parse_response_to_df(response_text: str) -> pd.DataFrame:
    """
    Parse response string thành DataFrame có cột StoreID, StoreName, BrandID.
    """
    rows = response_text.strip().split("\n")  # tách từng dòng
    parsed_data = []

    for row in rows:
        if "=" not in row:
            continue  # bỏ qua dòng không hợp lệ
        left, brand_id = row.split("=", 1)
        if "-" in left:
            store_id, store_name = left.split("-", 1)
        else:
            store_id, store_name = left, ""
        parsed_data.append([store_id.strip(), store_name.strip(), brand_id.strip()])

    return pd.DataFrame(parsed_data, columns=["StoreID", "StoreName", "BrandName"])
````
