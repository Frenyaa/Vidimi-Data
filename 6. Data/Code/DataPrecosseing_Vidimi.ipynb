```python
# Drop the specified columns
columns_to_drop = ['Vendor Brand ID', 'Spend ID', 'Original Brand Id ', 'Item Category Id', 'Age ID', 'Gender ID']
df.drop(columns=columns_to_drop, inplace=True)

# Display the info of the DataFrame after dropping columns
df.info()
```

```python
import pandas as pd

def missing_values_table(df):
    # Tính số lượng và tỷ lệ missing values
    mis_val = df.isnull().sum()
    mis_val_percent = 100 * mis_val / len(df)

    # Gộp hai cột lại thành bảng
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1)
    mis_val_table.columns = ['Missing Values', '% of Total Values']

    # Lọc các cột có missing values > 0
    mis_val_table_filtered = (
        mis_val_table[mis_val_table['Missing Values'] > 0]
        .sort_values('% of Total Values', ascending=False)
        .round(1)
    )

    print(
        f"Your selected dataframe has {df.shape[1]} columns.\n"
        f"There are {mis_val_table_filtered.shape[0]} columns that have missing values."
    )

    # Lọc các cột không có missing values
    no_missing_cols = mis_val_table[mis_val_table['Missing Values'] == 0].index.tolist()

    if no_missing_cols:
        print("\nCột không bị missing value:", ", ".join(no_missing_cols))
    else:
        print("\nKhông có cột nào đầy đủ dữ liệu (tất cả đều có missing value).")

    print("\nBảng các cột có missing value:")
    display(mis_val_table_filtered)

    return mis_val_table_filtered
```

```python
# Drop rows with missing values in the specified columns
df.dropna(subset=['Nganh Hang ID', 'Khu vực ID', 'Item Name', 'Item Number', 'Item Price'], inplace=True)

```

2.3.Xử lý các hóa đơn nghi ngờ
---

- Thông qua file ggcolab DetectAnomoly mà chúng ta xác định được có những trường hợp spam bill

- Rule để xác định đó là spam bill

Cùng userID cùng PlaceID và cùng trong một ngày

-> Hướng xử lý

Giữ lại một bill, những bill còn lại xóa đi

```python
import pandas as pd

# Đọc lại file những bill bị đánh giá gian lận
file_path = '/content/fraud_cases (1).xlsx'

# You might need to adjust the read function based on the actual file type

fc = pd.read_excel(file_path)
display(fc.head(2))
```

```python
# Extract the Bill IDs to be removed from the 'bill_problem' column in fc
bill_ids_to_remove = []
for index, row in fc.iterrows():
    bill_list = eval(row['bill_problem']) # Use eval to convert the string representation of a list to a list
    if len(bill_list) > 1:
        bill_ids_to_remove.extend(bill_list[1:])

# Remove rows from df where 'Bill ID' is in the list of bill_ids_to_remove
df = df[~df['Bill ID'].isin(bill_ids_to_remove)]

# Display the first few rows of the modified DataFrame
display(df.head())
```

```python
# Count the number of unique values in the 'Bill ID' column
unique_bill_ids_count = df['Bill ID'].nunique()

# Display the count
print(f"Number of unique Bill IDs: {unique_bill_ids_count}")
```
