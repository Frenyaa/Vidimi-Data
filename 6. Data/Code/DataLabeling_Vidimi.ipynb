2.2.Tách chữ ra để có dữ liệu train

```python
import re

def extract_item_names(items_string):
    if pd.isna(items_string):
        return ""
    # Split the string by newline characters
    items_list = items_string.split('\n')
    extracted_names = []
    for item in items_list:
        # Find the part after "name: " and before ", slg"
        match = re.search(r'name: (.*?),\s*slg', item)
        if match:
            extracted_names.append(match.group(1).strip())
        else:
            # If the pattern doesn't match, try to extract just the name part
            name_match = re.search(r'name: (.*)', item)
            if name_match:
                extracted_names.append(name_match.group(1).strip())

    # Join the extracted names with a newline character
    return '\n'.join(extracted_names)

# Apply the function to the 'Items' column of the filtered DataFrame
# Use .loc to avoid SettingWithCopyWarning
df_filtered.loc[:, 'Extracted_Items'] = df_filtered['Items'].apply(extract_item_names)

# Display the head of the DataFrame with the new column
display(df_filtered[['Items', 'Extracted_Items']].head())
```

2.3. Kết nối các dữ liệu để thành prompt cho gemini

```python
# Create a new DataFrame with 'Bill ID' and 'Extracted_Items' columns
df_for_gemini = df_filtered[['Bill ID', 'Extracted_Items']].copy()

# Format the data into the desired string format for each row
def format_for_gemini(row):
    bill_id = row['Bill ID']
    extracted_items = row['Extracted_Items']
    # Split the items string by newline and join with newline again
    formatted_items = extracted_items.replace('\n', '\n')
    return f"*{bill_id}\n{formatted_items}"

# Apply the formatting function to each row to create a new column
df_for_gemini['Formatted_Data'] = df_for_gemini.apply(format_for_gemini, axis=1)

# Display the head of the DataFrame with the new formatted column
display(df_for_gemini.head())
```

```python
# Define the chunk size
chunk_size = 410
# 30 vẫn còn chạy tốt khi wait time là 60s

# Create a list to store the combined strings
formatted_data_chunks = []

# Iterate through the DataFrame in chunks
for i in range(0, len(df_for_gemini), chunk_size):
    # Select the chunk of rows
    chunk = df_for_gemini.iloc[i:i + chunk_size]
    # Join the 'Formatted_Data' column of the chunk with newline characters
    combined_data = '\n'.join(chunk['Formatted_Data'])
    # Append the combined data to the list
    formatted_data_chunks.append(combined_data)

# Display the first few combined chunks as an example
for i, chunk in enumerate(formatted_data_chunks[:2]):
    print(f"--- Chunk {i+1} ---")
    print(chunk)
    print("-" * 20)
```

```python
# Đây là cái chạy cho nhập chunk bắt đầu và chunk kết thúc
# Nhập phạm vi chunk muốn chạy (tính từ 1 cho dễ nhìn)
start_chunk = int(input("Nhập số chunk bắt đầu: "))
end_chunk = int(input("Nhập số chunk kết thúc: "))

# Đảm bảo không vượt ngoài phạm vi
start_chunk = max(1, start_chunk)
end_chunk = min(len(formatted_data_chunks), end_chunk)


# Duyệt qua range cần chạy (chú ý: index trong Python bắt đầu từ 0)
for i in range(start_chunk-1, end_chunk):
    try:
        print(f"Processing chunk {i+1}/{len(formatted_data_chunks)}")
        # Combine the prompt and the data chunk
        full_prompt = prompt + "\n" + formatted_data_chunks[i]
        response = client.models.generate_content(
            model="gemini-2.0-flash-lite",
            contents=full_prompt,
            config=types.GenerateContentConfig(
                thinking_config=types.ThinkingConfig(thinking_budget=0)
            ),
        )

        # Append the model's response to the list
        gemini_responses.append(response.text)

        # 🔹 Chỉ in ra response của chunk hiện tại
        print(f"\n--- Response for chunk {i+1} ---")
        print(response.text)
        print("-" * 20)

        # Pause 60s before next request
        if i < end_chunk - 1:
            print("⏳ Waiting 60s to avoid rate limit...")
            time.sleep(60)

    except Exception as e:
        print(f"Error processing chunk {i+1}: {e}")
        gemini_responses.append(f"Error processing chunk {i+1}: {e}")
```

```python
import pandas as pd
import re

def parse_receipt_blocks(data_list):
    parsed_data = []

    # Ghép list thành chuỗi rồi tách theo dấu *
    data_str = "".join(data_list)
    blocks = data_str.split("*")

    for block in blocks:
        block = block.strip()
        if not block:
            continue

        # Lấy BillID và IndusLabel từ dòng đầu
        header_match = re.match(r'([0-9a-f]+)\s*-\s*(\d+)', block)
        if not header_match:
            continue
        bill_id = header_match.group(1)
        indus_label = header_match.group(2)

        # Tách các dòng sản phẩm (bỏ dòng đầu header)
        lines = block.split("\n")[1:]
        for line in lines:
            line = line.strip()
            if not line:
                continue
            # Sản phẩm dạng "Tên - Sector"
            prod_match = re.match(r'^(.*?)\s*-\s*([\d\.]+)$', line)
            if prod_match:
                product = prod_match.group(1).strip()
                sector = prod_match.group(2)
                parsed_data.append([bill_id, product, indus_label, sector])

    df = pd.DataFrame(parsed_data, columns=["BillID", "Product", "IndusLabel", "SectorLabel"])
    return df

```
